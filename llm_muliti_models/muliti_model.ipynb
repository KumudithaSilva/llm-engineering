## Load Libraries

```python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

## Load Environment Key"

```python
try:
    script_dir = os.path.dirname(os.path.abspath(__file__))
except NameError:
    script_dir = os.getcwd()

# Go up one level to the project folder
project_dir = os.path.dirname(script_dir)
env_path = os.path.join(project_dir, "env_keys", ".env")

# Access the variable
load_dotenv(dotenv_path=env_path)

openai_api_key = os.getenv("OPENAI_API_KEY")
google_api_key = os.getenv("GOOGLE_API_KEY")
gork_api_key = os.getenv("GROK_API_KEY")
open_route_api_key = os.getenv("OPRN_ROUTER_KEY")

print("API Keys loaded:")
print(f"OpenAI API Key: {openai_api_key is not None}")
print(f"Google API Key: {google_api_key is not None}")
print(f"Grok API Key: {gork_api_key is not None}")
print(f"OpenRoute API Key: {open_route_api_key is not None}")
```

```python
# Connect to OpenAI client library
openai = OpenAI()

gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
grok_url = "https://api.x.ai/v1"
ollama_url = "http://localhost:11434/v1"
openrouter_url = "https://openrouter.ai/api/v1"

# Initialize clients for different language models
gemini = OpenAI(api_key=google_api_key, base_url=gemini_url)
grok = OpenAI(api_key=gork_api_key, base_url=grok_url)
ollama = OpenAI(api_key="ollama", base_url=ollama_url)
openrouter = OpenAI(base_url=openrouter_url, api_key=open_route_api_key)
```

```python
tell_a_joke = [
    {
        "role": "user",
        "content" :  "Tell a joke for student who are new to Calculus"
    }
]
```

```python
response = openai.chat.completions.create(model="gpt-4.1-nano", messages=tell_a_joke)
display(Markdown(response.choices[0].message.content))
```

```python
response = gemini.chat.completions.create(model="gemini-2.5-flash", messages=tell_a_joke)
display(Markdown(response.choices[0].message.content))
```
