## Define the prompt for the conversation

```python
gpt_model = "gpt-4.1-mini"
gemini_model = "gemini-2.5-flash"
```

```python
gpt_chat = "You are a chatbot who is very polite; \
you likes nature and asked questings regarding nature."

gemini_chat = "You are a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is polite, \
you try to calm more and keep chatting."
```

```python
gpt_messages = "Hi there"
gemini_messages = "Hi"
```

```python
def call_gpt():
    message = [{"role": "system","content": gpt_chat}]
    for gpt, gemini in zip(gpt_messages, gemini_messages):
        message.append({"role": "assistant","content": gpt})
        message.append({"role": "user","content": gemini})
    response = openai.chat.completions.create(model="gpt-4.1-mini", messages=message)
    return response.choices[0].message.content

starter = call_gpt()
starter
```

```python
def call_gemini():
    message = [{"role": "system","content": gemini_chat}]
    message.append({"role": "assistant","content": gemini_messages})
    message.append({"role": "user", "content": starter})
    response = openai.chat.completions.create(model="gpt-4.1-mini", messages=message)
    return response.choices[0].message.content

call_gemini()
```
