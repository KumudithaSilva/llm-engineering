### Hugging Face Tokenizer

```python
from transformers import AutoTokenizer
```

```python
# tokenizer = AutoTokenizer.from_pretrained("meta-llama/Meta-Llama-3.1-8B")
tokenizer = AutoTokenizer.from_pretrained("tiiuae/falcon-7b")
tokens = tokenizer.encode("We are learning about tokenization in HuggingFace Transformers library.")
print(tokens)
```

### Instruct variants of models

```python
tokenizer_instruct = AutoTokenizer.from_pretrained("tiiuae/falcon-7b-instruct")
```

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Tell users regarding HuggingFace Transformers libraries."}
  ]

prompt = tokenizer_instruct.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
print(prompt)
```

### Instruct variants of other models

```python
PHI4 = "microsoft/Phi-4-mini-instruct"
DEEPSEEK = "deepseek-ai/DeepSeek-V3.1"
QWEN_CODER = "Qwen/Qwen2.5-Coder-7B-Instruct"
```

```python
phi4_tokenizer = AutoTokenizer.from_pretrained(PHI4)

text = "Tell users regarding HuggingFace Transformers libraries."
print("\nPhi 4:")
tokens = phi4_tokenizer.encode(text)
print(tokens)
print(phi4_tokenizer.batch_decode(tokens))
```

```python
deepseek_tokenizer = AutoTokenizer.from_pretrained(DEEPSEEK)

text = "Tell users regarding HuggingFace Transformers libraries."
print("\nDeepseek:")
tokens = (deepseek_tokenizer.encode(text))
print(tokens)
print(deepseek_tokenizer.batch_decode(tokens))
```

```python
qwencoder_tokenizer = AutoTokenizer.from_pretrained(QWEN_CODER)

text = "Tell users regarding HuggingFace Transformers libraries."
print("\nQwen:")
tokens = (qwencoder_tokenizer.encode(text))
print(tokens)
print(qwencoder_tokenizer.batch_decode(tokens))
```

```python
qwen_tokenizer = AutoTokenizer.from_pretrained(QWEN_CODER)
code = """
def hello_world(person):
  print("Hello", person)
"""
tokens = qwen_tokenizer.encode(code)
for token in tokens:
  print(f"{token}={qwen_tokenizer.decode(token)}")
```
